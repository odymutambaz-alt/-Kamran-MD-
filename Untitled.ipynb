```python


const ("OWNER_NUMBER ")= "26377247017"
const BOT_NAME = "KAMRAN-MD"
let BAN_LIST = []

function isOwner(number){ return number === OWNER_NUMBER }
function isBanned(number){ return BAN_LIST.includes(number) }

async function startBot() {
    const { state, saveCreds } = await useMultiFileAuthState('session')
    const sock = makeWASocket({
        auth: { creds: state.creds, keys: makeCacheableSignalKeyStore(state.keys, pino().child({ level: "silent" })) },
        logger: pino({ level: "silent" }),
        printQRInTerminal: false,
        browser: [BOT_NAME, "Chrome", "1.0"]
    })

    if (!sock.authState.creds.registered) {
        console.log(`\nWaiting for pairing... Owner: ${OWNER_NUMBER}`)
        await new Promise(r => setTimeout(r, 3000))
        let code = await sock.requestPairingCode(OWNER_NUMBER)
        console.log(`\n╔════════════════════════════╗`)
        console.log(`║ PAIRING CODE: ${code}  ║`)
        console.log(`╚════════════════════════════╝`)
        console.log(`\nWhatsApp > Linked Devices > Link with phone number > Enter code\n`)
    }

    sock.ev.on("creds.update", saveCreds)

    sock.ev.on("messages.upsert", async (m) => {
        try {
            let msg = m.messages[0]
            if (!msg.message || msg.key.fromMe) return
            let senderJid = msg.key.remoteJid
            let senderNumFull = msg.key.participant || senderJid
            let number = senderNumFull.replace(/@.*/, "").replace("+", "").replace(" ", "")
            let text = msg.message.conversation || msg.message.extendedTextMessage?.text || ""
            let isGroup = senderJid.endsWith("@g.us")
            let isGroupAdmin = false
            if (isGroup) {
                try {
                    let meta = await sock.groupMetadata(senderJid)
                    isGroupAdmin = !!meta.participants.find(p => p.id === senderNumFull && p.admin)
                } catch {}
            }

            // Python logic ported: BAN CHECK
            if (isBanned(number)) {
                await sock.sendMessage(senderJid, { text: `❌ @${number} banned from ${BOT_NAME}` })
                return
            }

            if (text.toLowerCase() === ".ping") {
                await sock.sendMessage(senderJid, { text: `⚡ ${BOT_NAME} Online!\nOwner: ${OWNER_NUMBER}\nUser: ${number}` })
            }

            if (text.startsWith(".ban")) {
                if (!isOwner(number)) return sock.sendMessage(senderJid, { text: "❌ Owner only!" })
                let target = text.split(" ")[1]?.replace(/[^0-9]/g, "")
                if (target) {
                    if (!BAN_LIST.includes(target)) BAN_LIST.push(target)
                    await sock.sendMessage(senderJid, { text: `✅ Banned ${target}\nBAN_LIST = [${BAN_LIST.join(", ")}]` })
                } else {
                    await sock.sendMessage(senderJid, { text: "Use: .ban 263771234567" })
                }
            }

            if (text.startsWith(".unban")) {
                if (!isOwner(number)) return
                let target = text.split(" ")[1]?.replace(/[^0-9]/g, "")
                BAN_LIST = BAN_LIST.filter(n => n !== target)
                await sock.sendMessage(senderJid, { text: `✅ Unbanned ${target}` })
            }

            if (text === ".banlist") {
                if (!isOwner(number)) return sock.sendMessage(senderJid, { text: "❌ Owner only!" })
                await sock.sendMessage(senderJid, { text: `📋 Banned: ${BAN_LIST.length ? BAN_LIST.join(", ") : "None"}` })
            }

            if (text.startsWith(".kick") || text.startsWith(".remove")) {
                if (!isGroup) return
                if (!isGroupAdmin && !isOwner(number)) return sock.sendMessage(senderJid, { text: "❌ Need admin!" })
                let mentioned = msg.message.extendedTextMessage?.contextInfo?.mentionedJid?.[0]
                let target = mentioned || (text.split(" ")[1]?.replace(/[^0-9]/g, "") + "@s.whatsapp.net")
                if (target && target !== "@s.whatsapp.net") await sock.groupParticipantsUpdate(senderJid, [target], "remove")
            }

            if (text === ".help" || text === ".menu") {
                await sock.sendMessage(senderJid, { text: `*${BOT_NAME} Commands:*\n.ping - Check online\n.help - Menu\n.ban 263... - Ban from bot (owner)\n.unban 263... - Unban (owner)\n.kick @user - Kick from group (admin)\n.banlist - Show banned (owner)\n\nOwner: ${OWNER_NUMBER}` })
            }

        } catch (e) { console.log("Error:", e) }
    })
}
startBot()
```
